```python
bike_num = (df['current_num']/df['total_num']).value_counts()
```

```python
# bike_num
proportion = {'lower than 10%' : 0, 'between 10% & 20%' : 0, 'between 20% & 30%' : 0,
              'between 30% & 40%' : 0, 'between 40% & 50%' : 0,
              'between 50% & 60%' : 0, 'between 60% & 70%' : 0, 'between 70% & 80%' : 0,
              'between 80% & 90%' : 0, 'between 90% & 100%' : 0}
for i in range(len(bike_num.index)):
  if bike_num.index[i] <= (10 / 100):
    proportion['lower than 10%'] = proportion['lower than 10%'] + bike_num.values[i]
    continue
  elif bike_num.index[i] <= (20 / 100):
    proportion['between 10% & 20%'] = proportion['between 10% & 20%'] + bike_num.values[i]
    continue
  elif bike_num.index[i] <= (30 / 100):
    proportion['between 20% & 30%'] = proportion['between 20% & 30%'] + bike_num.values[i]
    continue
  elif bike_num.index[i] <= (40 / 100):
    proportion['between 30% & 40%'] = proportion['between 30% & 40%'] + bike_num.values[i]
    continue
  elif bike_num.index[i] <= (50 / 100):
    proportion['between 40% & 50%'] = proportion['between 40% & 50%'] + bike_num.values[i]
    continue
  elif bike_num.index[i] <= (60 / 100):
    proportion['between 50% & 60%'] = proportion['between 50% & 60%'] + bike_num.values[i]
    continue
  elif bike_num.index[i] <= (70 / 100):
    proportion['between 60% & 70%'] = proportion['between 60% & 70%'] + bike_num.values[i]
    continue
  elif bike_num.index[i] <= (80 / 100):
    proportion['between 70% & 80%'] = proportion['between 70% & 80%'] + bike_num.values[i]
    continue
  elif bike_num.index[i] <= (90 / 100):
    proportion['between 80% & 90%'] = proportion['between 80% & 90%'] + bike_num.values[i]
    continue
  elif bike_num.index[i] <= (100 / 100):
    proportion['between 90% & 100%'] = proportion['between 90% & 100%'] + bike_num.values[i]
    continue
```
